## 6. Simple Calculator


```python
def simple_calculator(num1, operator, num2):
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero is undefined!"
    elif operator == "%":
        if num2 != 0:
            return num1 % num2
        else:
            return "Error: Modulo by zero is undefined!"
    else:
        return f"Error: Invalid operator '{operator}'!"
test_cases = [
    (15, "+", 25),
    (50, "-", 18),
    (12, "*", 8),
    (100, "/", 4),
    (42, "/", 0),
    (17, "%", 5),
    (10, "^", 2)
]
print("=== CALCULATOR TEST RUN ===")
for n1, op, n2 in test_cases:
    res = simple_calculator(n1, op, n2)
    print(f"{n1:5.1f} {op} {n2:5.1f} = {res}")
```

## 7. Random Numbers and Even Number Filter


```python
import random
random.seed(42)
random_numbers = [random.randint(1, 100) for _ in range(10)]
print("Generated Random Numbers:", random_numbers)
even_numbers_loop = []
for num in random_numbers:
    if num % 2 == 0:
        even_numbers_loop.append(num)
even_numbers_comp = [num for num in random_numbers if num % 2 == 0]
print("Even Numbers (via loop) :", even_numbers_loop)
print("Even Numbers (via comp) :", even_numbers_comp)
odd_numbers = [num for num in random_numbers if num % 2 != 0]
print("Odd Numbers             :", odd_numbers)
```

## 8. String and Word Reversal


```python
original_text = "Python Programming is Fun"
print("Original String:               ", original_text)
reversed_string = original_text[::-1]
print("Reversed Complete String:      ", reversed_string)
words = original_text.split()
words_reversed_order = " ".join(words[::-1])
print("Reversed Word Order:           ", words_reversed_order)
each_word_reversed = " ".join(word[::-1] for word in words)
print("Each Individual Word Reversed: ", each_word_reversed)
assignment_text = "Python is easy to learn"
print("\nAssignment Sample String:         ", assignment_text)
print("Reversed Complete String:         ", assignment_text[::-1])
reversed_assignment_words = " ".join(assignment_text.split()[::-1])
print("Reversed Word Order:              ", reversed_assignment_words)
```

## 9. Basic NumPy and Matplotlib


```python
import numpy as np
import matplotlib.pyplot as plt
array1 = np.array([1, 2, 3, 4, 5])
print("Original NumPy Array :", array1)
squared_array = array1 ** 2
print("Squared Array (x^2)  :", squared_array)
years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
adoption_index = [10, 18, 25, 38, 52, 65, 78, 88, 95, 100]
plt.figure(figsize=(9, 4.5), dpi=100)
plt.plot(years, adoption_index, marker='o', color='#1f77b4', linewidth=2.5, markersize=7, label="AI Adoption Rate")
plt.title("Growth of Artificial Intelligence Adoption (2015 - 2024)", fontsize=13, pad=12)
plt.xlabel("Year", fontsize=11)
plt.ylabel("Adoption Index (0-100)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
