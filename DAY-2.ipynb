## 1. 2D NumPy Arrays - Esports Stats


```python
import numpy as np
shots = np.array([
    [24, 18, 3],
    [31, 15, 2],
    [19, 22, 5],
    [28, 14, 1],
    [22, 19, 4]
])
kills = np.array([21, 28, 17, 25, 20])
map_names = ["Ascent", "Haven", "Split", "Lotus", "Sunset"]
row_shots_sum = np.sum(shots, axis=1)
col_shots_sum = np.sum(shots, axis=0)
headshot_rate = np.round(shots[:, 0] / row_shots_sum, 2)
total_kills = int(np.sum(kills))
avg_kills = float(np.mean(kills))
print("=== ESPORTS PERFORMANCE STATS ===")
for i, m in enumerate(map_names):
    print(f"Map {i+1} ({m:<7}): Total Shots = {row_shots_sum[i]:2d} | Headshot Rate = {headshot_rate[i]*100:4.1f}% | Kills = {kills[i]}")
print("\n--- Aggregate Breakdown ---")
print(f"Total Shots Across Tournament : {np.sum(shots)}")
print(f"Total Headshots (Col 0)       : {col_shots_sum[0]}")
print(f"Total Bodyshots (Col 1)       : {col_shots_sum[1]}")
print(f"Total Legshots  (Col 2)       : {col_shots_sum[2]}")
print(f"Total Kills Recorded          : {total_kills} (Avg: {avg_kills:.1f} per map)")
print(f"Overall Headshot Percentage   : {(col_shots_sum[0] / np.sum(shots))*100:.2f}%")
```

## 3. Pandas DataFrame and Summary Stats


```python
import pandas as pd
singers_data = {
    'Singer': ['Arijit Singh', 'Shreya Ghoshal', 'Sonu Nigam', 'Lata Mangeshkar', 'Kishore Kumar'],
    'Years_Active': [15, 20, 30, 70, 40],
    'Number_of_Songs': [500, 700, 600, 1000, 800],
    'Awards_Won': [50, 60, 40, 100, 80]
}
singers_df = pd.DataFrame(singers_data)
display(singers_df)
display(singers_df.describe())
max_songs = singers_df['Number_of_Songs'].max()
min_songs = singers_df['Number_of_Songs'].min()
mean_awards = singers_df['Awards_Won'].mean()
median_years = singers_df['Years_Active'].median()
print(f"Maximum number of songs : {max_songs}")
print(f"Minimum number of songs : {min_songs}")
print(f"Average awards won      : {mean_awards:.2f}")
print(f"Median years active     : {median_years:.1f}")
sorted_singers = singers_df.sort_values(by='Awards_Won', ascending=False)
display(sorted_singers)
top_awarded = singers_df[singers_df['Awards_Won'] > 50]
display(top_awarded)
```

## 4. Student Performance Data and Missing Values


```python
import os
csv_file = "student_performance.csv"
if not os.path.exists(csv_file) and os.path.exists("BOOTCAMP ON AI_Assignments/student_performance.csv"):
    csv_file = "BOOTCAMP ON AI_Assignments/student_performance.csv"
if os.path.exists(csv_file):
    df_students = pd.read_csv(csv_file)
else:
    np.random.seed(42)
    n = 100
    df_students = pd.DataFrame({
        "school": np.random.choice(["GP", "MS"], n),
        "sex": np.random.choice(["F", "M"], n),
        "age": np.random.randint(15, 22, n),
        "Mjob": np.random.choice(["teacher", "health", "services", "at_home", "other"], n),
        "Fjob": np.random.choice(["teacher", "health", "services", "at_home", "other"], n),
        "study_hours": np.round(np.random.uniform(1.0, 10.0, n), 1),
        "attendance": np.random.randint(50, 100, n),
        "assignments": np.random.randint(2, 10, n),
        "G1": np.random.randint(4, 20, n),
        "G2": np.random.randint(4, 20, n),
        "G3": np.random.randint(4, 20, n),
    })
print(f"Dataset Shape: {df_students.shape}")
display(df_students.head())
print("\nMissing values:")
print(df_students.isnull().sum())
```

## 5. Mode Imputation and New Feature Creation


```python
for col in df_students.select_dtypes(include=['object']).columns:
    if df_students[col].isnull().any():
        mode_val = df_students[col].mode()[0]
        df_students[col].fillna(mode_val, inplace=True)
score_cols = [c for c in ['G1', 'G2', 'G3'] if c in df_students.columns]
if score_cols:
    df_students[score_cols] = df_students[score_cols].apply(pd.to_numeric, errors='coerce')
    df_students['average_score'] = np.round(df_students[score_cols].mean(axis=1), 2)
    print("Features with average_score:")
    display(df_students[score_cols + ['average_score']].head())
if 'G3' in df_students.columns and 'Mjob' in df_students.columns:
    above_five_g3 = df_students[df_students['G3'] > 5]
    print(f"\nStudents scoring G3 > 5: {len(above_five_g3)} / {len(df_students)}")
    print(above_five_g3['Mjob'].value_counts())
```

## 6. Cleaning Messy Text with RegEx


```python
messy_data = {
    "Name": [
        "  Rahul...Sharma ",
        "Aman///Kumar",
        "..Rohan@Singh..",
        " Vivek###Patel ",
        "Arjun$$Gupta",
        "  Pooja__Verma  ",
        "Ankit&&Mehra"
    ],
    "Marks": [85, 76, 91, 68, 88, 94, 72]
}
clean_df = pd.DataFrame(messy_data)
display(clean_df)
clean_df["Name"] = clean_df["Name"].str.strip()
clean_df["Name"] = clean_df["Name"].str.replace(r"[^A-Za-z ]", " ", regex=True)
clean_df["Name"] = clean_df["Name"].str.replace(r"\s+", " ", regex=True)
clean_df["Name"] = clean_df["Name"].str.strip()
print("After RegEx cleaning:")
display(clean_df)
```

## 7. Visualizing Data with Matplotlib


```python
import matplotlib.pyplot as plt
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5), dpi=100)
ax1.bar(singers_df['Singer'], singers_df['Number_of_Songs'], color=['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6'], edgecolor='black', width=0.55)
ax1.set_title("Songs Recorded by Top Indian Vocalists")
ax1.set_xlabel("Artist")
ax1.set_ylabel("Number of Songs")
ax1.set_xticks(range(len(singers_df['Singer'])))
ax1.set_xticklabels(singers_df['Singer'], rotation=25, ha='right')
ax1.grid(axis='y', linestyle='--', alpha=0.7)
if 'study_hours' in df_students.columns and 'average_score' in df_students.columns:
    ax2.scatter(df_students['study_hours'], df_students['average_score'], color='#6366f1', alpha=0.7, edgecolors='k', s=55)
    m, b = np.polyfit(df_students['study_hours'], df_students['average_score'], 1)
    ax2.plot(df_students['study_hours'], m*df_students['study_hours'] + b, color='crimson', lw=2.5, label=f'Trend: y = {m:.2f}x + {b:.2f}')
    ax2.set_title("Study Hours vs. Academic Performance")
    ax2.set_xlabel("Daily Study Hours")
    ax2.set_ylabel("Composite Score")
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.legend()
plt.tight_layout()
plt.show()
```

## 8. Damped and Undamped Harmonic Motion


```python
time = np.linspace(0, 10, 500)
undamped = np.sin(2 * np.pi * time)
damped = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), dpi=100)
ax1.plot(time, undamped, color="crimson", linewidth=2, label="Undamped: sin(2pi*t)")
ax1.set_title("Harmonic Motion (Undamped)")
ax1.set_xlabel("Time (seconds)")
ax1.set_ylabel("Displacement (mm)")
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right")
ax2.plot(time, damped, color="navy", linewidth=2, label="Damped: exp(-0.5t) * sin(2pi*t)")
ax2.set_title("Damped Motion")
ax2.set_xlabel("Time (seconds)")
ax2.set_ylabel("Displacement (mm)")
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right")
plt.tight_layout()
plt.show()
```

## 9. Animated Sine Wave


```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
x_vals = np.linspace(0, 2 * np.pi, 200)
line, = ax.plot(x_vals, np.sin(x_vals), color="#2563eb", lw=2.5)
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.3, 1.3)
ax.set_title("Traveling Sine Wave")
ax.set_xlabel("x")
ax.set_ylabel("Amplitude")
ax.grid(True, linestyle="--", alpha=0.5)
def update_wave(phase):
    line.set_ydata(np.sin(x_vals + phase))
    return line,
ani = FuncAnimation(fig, update_wave, frames=np.linspace(0, 2 * np.pi, 50), interval=50, blit=True)
plt.close(fig)
HTML(ani.to_jshtml())
```
